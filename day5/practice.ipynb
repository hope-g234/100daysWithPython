```python
studen_scores = [85, 92, 78, 90, 88]

total_mark = sum(studen_scores)

total = 0
for score in studen_scores:
    total += score

print(total)

# max
max_mark = 0
for mark in studen_scores:
    if mark > max_mark:
        max_mark = mark

print(max_mark)

#min
min_mark = studen_scores[0]
for mark in studen_scores:
    if mark <= min_mark:
        min_mark = mark
print(min_mark)

```

```python
#FizzBuzz
for number in range(1,101):
    if number % 3 == 0 and number % 5 == 0:
        print("FizzBuzz")
    elif number % 3 == 0:
        print("Fizz")
    elif number % 5 == 0:
        print("Buzz")
    else:
        print(number)
```
