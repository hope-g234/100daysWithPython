```python
# learning about scope
"""  Local and global variables """
weight = 50

def bmi():
    weight = 20
    print(weight)
# can call global weight inside the function but don't modify it since it's  confusing
bmi()
print(weight)
```

```python
"""there is no block scope in python - in python there is no block scope.Inside a if/else/for/while code block is the same as outside it """
"""modify variables with global scope will create bug and error"""
numbers = 1

def increase_number(number):
    print(f"numbers inside function:{numbers}")
    return number+1

numbers = increase_number(numbers)
print(numbers)
```
