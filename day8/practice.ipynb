```python
def greet():
    print("Hello")
    print("Isn't the weather nice!")

greet()

def greet_with_name(name):
    print(f"Hello, {name}!")
    print("Isn't the weather nice!")

greet_with_name("Angela")
```

```python
def greet_with(name,location):
    print(f"Hello, {name}!")
    print(f"what is it like {location}?")

greet_with(name="jack",location="rome")
```
